# 22. 作業化系統：STEP、OEF 與機率溝通

{doc}`上一章 <21_psha>`把預報接到了危害度：短期地震率餵進 PSHA
的積分，就得到會隨序列演化的地動超越機率。那是整條鏈的**輸出
端**。這一章往回走，看這條鏈的**中段**——一個統計模型要變成
一台 24 小時運轉、對社會發布數字的機器，中間要補上哪些東西。

引擎我們早就看過了。{doc}`第 14 章 <14_etas_estimation>`的 14.7
節從 ETAS 推出 Reasenberg–Jones 的解析特例 {eq}`eq:rj-rate`，
並列出加州通用參數、加州實測前震率、日本參數三者的對照——同一
條式子換一組區域參數，答案差一倍以上。那條 1989 年的式子到今天
仍是多國作業系統的骨幹。本章不重推它，只問一件事：**把它接上
電源、讓它每天對外講話，還缺什麼**？

答案不太符合理工科的直覺。缺的不是更好的概似函數，而是：資料在
深夜三點壞掉時誰值班、機率算出來由誰簽名發布、5% 這個數字要用
什麼句型說給一個剛失去房子的人聽、以及當模型根本不存在時該怎麼
辦。這一章的主角因此有三個：**STEP**（Gerstenberger et al. 2005,
*Nature*）這個把「明天的地震」變成即時地圖的原型系統、三個國家
二十年來長出的三種制度，以及一門有假說、有實驗、有可測量結果的
科學——機率溝通。

## 22.1 從模型到系統的七步

先把整條路徑攤開。從一份地震目錄到一個能支撐工程決策的系統，中
間有七個台階，每一階都會冒出前一階沒有的問題：

| 步 | 這一步在做什麼 | 新冒出來的需求 |
|---|---|---|
| 1 經驗律 | Omori–Utsu、GR | 一份目錄 |
| 2 模型 | R–J、ETAS、STEP、EEPAS | 參數估計、$M_c$、要不要模擬 |
| 3 檢驗 | CSEP、前瞻、基準比較 | **一個社群**（第三方） |
| 4 系統 | 自動化、近即時、與國家危害模型整合 | 工程、機房、成本取捨 |
| 5 權威 | 誰有法定責任發布 | **制度** |
| 6 溝通 | 說給誰、怎麼說、多久一次 | 社會科學、共同設計 |
| 7 風險 | OELF、時變危害、補強政策 | 曝險與易損性資料 |

這張表的重點不在七這個數字，而在**第 3 步之後的斷層**。第 1、2
步是統計問題，可以一個人在筆電上做完。第 3 步開始需要別人——
因為「開發者自己相信自己的模型」在這個領域裡不構成證據
（{doc}`第 9 章 <09_forecasting_intro>`的 CSEP 三條紀律，工具箱
在{doc}`第 18 章 <18_testing_comparison>`）。而從第 4 步起，**瓶頸
就不再是統計**：義大利有全世界最完整的作業化系統，卻至今不對
大眾公開；紐西蘭因為 2012 年換了定位系統，模型參數到現在還卡在
換系統之前。這些都不是概似函數能解決的問題。

學生最容易誤判的也正是這裡：以為「把模型做得更準」是唯一的進步
方向。本章接下來的每一節，都是這個誤判的反例。

順帶把一個非常容易混淆的時間軸釘住。**地震預警（EEW）**是
「破裂已經發生、地震波還在路上」的**真即時**問題——秒的尺度，
拚的是通訊比 S 波快。本章的 OEF 則是「破裂**尚未**發生」的
**近即時**問題——分鐘到天的尺度，「近」這個字誠實承認了不可
避免的處理延遲：目錄要更新、參數要重估、機率要算、還要有人
看過才發得出去。兩者的輸入、輸出、使用者與失敗模式全都不同，
唯一的共同點是都得在半夜自動運轉。

## 22.2 STEP 的架構

STEP（Short-Term Earthquake Probability）2005 年在加州上線，是
第一個把即時餘震預報鋪成**空間地圖**的系統。它要補的兩個洞，
*Nature* 摘要親口點名：既有的通用短期叢集模型**不針對眼前這個
序列調校**，而且**不含餘震會發生在哪裡的資訊**。

### 三個複雜度遞增的元素

STEP 與 R–J、ETAS 的物理內核相同（疊加的 Omori 序列，每個地震
都能生自己的餘震），但架構不同：它不是一個自洽的單一模型，而是
**數個可以各自獨立視為模型的元件之組合**。時變的部分有三個元件，
複雜度遞增，切換由**資料量**決定：

1. **資料不足時**（序列剛開始，手上只有主震）：直接套用由過去
   地震活動導出的**區域通用參數**。空間上先以主震為中心給一個
   隨距離遞減的圓形餘震區，等真的有餘震進來，再據以估計斷層跡、
   修正餘震分布的形狀。
2. **當序列累積到完整規模以上至少 100 個餘震**：改用**該序列
   自己擬合**的參數。注意重新擬合會連帶改變整個序列的 $b$ 值，
   不只是 Omori 那幾個參數。
3. **當序列的生產力足夠高**（大到每個子區都還有足夠事件可擬合）：
   參數進一步**容許在空間上變化**，序列的不同段落各有各的行為。

三個元件依各自**對觀測資料的擬合程度**，以 AIC 加權合併
（Steacy et al. 2014 對 STEP 的描述即為此）。這裡只講觀念不寫
公式：**資料少時通用元件主導，資料累積後序列自身的行為接手**。
把這個切換畫成示意圖：

這張圖是本章第一個要記住的機制：**權重不是人訂的，是資料決定
的**。序列剛發生時，系統對它一無所知，只能借用整個區域的平均
行為；隨著餘震進來，這個序列自己的個性逐漸把通用先驗擠開。這
與下一節美國系統的貝氏設計是同一件事的兩種寫法——一個用模型
權重表達，一個用後驗分布表達。

### primary 與 secondary：誰是誰的餘震

STEP 把事件分成兩類。**primary** 是序列的母事件；**secondary**
則是某個前序事件的餘震——而那個前序事件本身不必是 primary。
連結的規則有兩條：餘震透過時空定義的「餘震區」歸屬到母事件，
而且一個事件**只能是規模相等或更大事件的餘震**。

這個區分有一個很實際的後果：**序列自身擬合的參數只用在 primary
事件的餘震率上，secondary 一律回到通用參數**。理由是資料量——
一個 secondary 事件自己的後代通常少到不夠擬合，硬要擬合只會得到
一組被雜訊主宰的參數。這是作業化系統裡反覆出現的取捨：**能不能
估**，跟**該不該估**，是兩個問題。

### 為什麼是「取最大」而不是相加

單一元件之內，某個位置的預報率取「所有前序 primary／secondary
事件在該位置預報的餘震率的**最大值**」，而不是把它們加起來。
這個規則叫 **Avmax**，後來被紐西蘭的混成危害模型整套沿用。

理由值得停下來想清楚，因為它是全章最容易搞混的一點。判準只有
一句話：**兩個率描述的是同一群事件，還是不同群事件**。

- 同一個位置附近的多個母事件，各自的餘震區重疊——它們**對同一
  塊地的同一批未來地震給出彼此競爭的估計**。相加等於重複計數，
  所以取最大（取包絡）。
- 時變叢集項與時間獨立背景項則不同：背景項描述的是「不屬於任何
  已知序列的地震」，叢集項描述的是「被已知事件觸發的地震」。
  兩者是**互斥的事件來源**，所以 STEP 對它們是**相加**。

同一章裡同時出現取最大與相加，不是不一致，是因為判準本來就在
「同一群 vs 不同群」上。附錄把這件事寫成兩行不等式。

時變項疊在時間獨立背景項上，還帶來一個結構性的好處：背景項通常
**直接接國家地震危害模型**（去叢集後的目錄平滑而得），因此當序列
衰減完畢、時變項歸零，整張地圖會自動退回長期危害模型。短期預報
與長期危害**天然一致**，不必事後對帳。這正是 21.5 節時變危害那條
曲線「整條上移、再整條落回」的模型端來源。

### STEP 與 ETAS 的取捨

| 面向 | STEP | ETAS |
|---|---|---|
| 計算 | 解析、便宜 | 未來事件的後代須模擬（萬次量級） |
| 二次觸發 | **部分**納入（靠 secondary 機制） | **完整**遞迴 |
| 架構 | 多元件組合，易與 NSHM 整合 | 自洽的單一模型 |
| 彈性 | 可換空間核、可吃有限斷層輸入 | 核形式較固定 |

這張表沒有贏家。ETAS 在統計上更誠實
（{doc}`第 13 章 <13_etas_structure>`的世代分解說明了為什麼
二次觸發不能省），STEP 在工程上更好養（第 4 步的成本取捨，
見 22.1 節）。義大利兩個都用，讓
集成的權重去決定；美國兩個都有，境內用便宜的、境外用貴的。

## 22.3 三國系統完整對照

二十年下來，世界上長出三種代表性的作業化路線。
{doc}`第 9 章 <09_forecasting_intro>`把這場對照留給了本章，因為
它要說的重點需要工程細節撐著才成立——三個系統算機率的能力相近，
差異幾乎全在**制度與資料**，不在模型。

| 面向 | 義大利 INGV | 紐西蘭 GNS/GeoNet | 美國 USGS |
|---|---|---|---|
| 引擎 | ETAS＋ETES＋STEP 集成 | 短期 STEP/ETAS、中期 EEPAS、長期 PPE | 境內 R–J 貝氏；境外 ETAS |
| 組合方式 | 依表現加權，**每週更新權重** | **Avmax** 逐點取最大 | 單一引擎，無集成 |
| 空間解析 | 0.1°×0.1°（排除火山區） | CSEP 紐西蘭測試區 | 事件頁；境外版可出圖 |

| 面向 | 義大利 INGV | 紐西蘭 GNS/GeoNet | 美國 USGS |
|---|---|---|---|
| 節奏 | 每日午夜＋每個 $M_L\ge3.5$ 之後，發布未來一週 | **無固定排程**，隨序列演化 | 大震後自動首報，之後按排程更新 |
| 公開 | **不對大眾公開**（民防、地方政府） | **完全公開**（網站＋社群媒體） | **完全公開**（事件頁） |
| 把關／痛點 | 候選模型**必須提交過 CSEP** | 目錄異質性；不預報 $M<5.0$ | 只貝氏更新生產力參數 |

三個系統各有一個值得單獨記住的細節。

### 義大利：把第三方檢驗寫進制度

嚴格按 Jordan et al.（2011）的定義，全世界**只有義大利**算完整
部署了 OEF：2013 年建立、每週產出預報、還含損失預報（OELF）。
它同時是最保守的——至今僅授權人員與民防體系可存取，近年才釋出
給地方政府討論如何運用。

最值得抄的是那條把關規則：**候選模型要加入官方集成，唯一的強制
條件是必須提交過 CSEP 實驗接受獨立透明的評估**。第 18 章那句
「開發者自己相信不算證據」，在義大利不是一句學術修辭，而是一道
行政門檻。另外兩個工程細節也值得知道：全程只用 $M_L$（明知它會
飽和，但為了目錄一致仍然只用它）；序列進行中**不更新生產力**，
模型在 2005–2009 的學習期校準完就固定。

### 紐西蘭：公開，以及目錄留下的疤

紐西蘭自 2010 年 M7.1 Darfield 地震起就對公眾發布預報，三個時間
尺度分工清楚：短期（時到週）是大震後的時間叢集，用 STEP、ETAS；
中期（月到年）是大震前的地震活動增加，用 EEPAS（第 16 章）；長期
（數十年）是空間叢集，用 PPE 與國家危害模型背景。三者以 Avmax
混成。

它的痛點是**目錄**。2012 年換了處理系統、規模尺標跟著偏移，導致
模型參數至今仍停在 2012 年之前的資料上，而且系統**不願預報
$M<5.0$**——因為小規模端的目錄異質性已經大到讓參數不可信。這對
任何有長期目錄異質性問題的地區（包括台灣）都是直接的警示：**你
的模型能預報到多小的規模，是你的觀測網歷史決定的，不是你的統計
方法決定的**。

### 美國：貝氏更新的具體化

美國境內是自動的 R–J 貝氏解，境外則由分析員人工啟動 ETAS 並跑
上萬次隨機事件集。境內系統的設計是「貝氏更新」最好的教材：
$t=0$ 時沒有任何本序列的資料，先驗是一個**涵蓋過去序列行為範圍**
的通用分布，因此預報**最寬**；隨著餘震進來，後驗逐步收斂到這個
序列自己的行為。**不確定性帶隨時間變窄，本身就是資訊**。

兩個誠實的細節。其一，目前只對**生產力參數**做貝氏更新，$c$
與 $p$ 固定在通用點估計，直到分析員判斷序列明顯偏離才放開——
三個參數只放開一個並不難理解：第 12 章的 12.3 節示範過，同一個
序列只要改變擬合窗長，$\hat p$ 就能從 0.9 走到 1.5 再崩到 0.7，
序列早期尤其如此。既然短窗估不穩，不如先釘住它，把資料留給最
有訊息的那個參數。通用模型則依構造環境分成十幾種，等於把「這是什麼
樣的地方」也編進先驗裡。其二，境內外系統**都不含背景地震活動
的預報**，只算餘震；在活躍地區、序列後期，這會低估真實危害。
順帶一提，「大震後約 20 分鐘自動發出首報」這個常被引用的數字，
出處是 Mizrahi et al.（2024）的回顧，不是一份系統規格書。

用合成資料把貝氏機制畫出來——先驗均值設在通用值，但這個序列
真實的生產力只有通用值的六成：

````
A_GEN, C_OM, P_OM = 30.0, 0.05, 1.08     # 通用參數下的餘震生產力（合成）
THETA_TRUE = 0.60                        # 本序列真實的生產力倍數
ALPHA0, BETA0 = 2.0, 2.0                 # 通用先驗 Gamma：均值 1、很寬

EDGES = np.array([0.0, 0.25, 0.5, 1.0, 3.0, 7.0, 30.0])
E_j = np.array([quad(lambda t: A_GEN * (t + C_OM) ** -P_OM, a, b)[0]
                for a, b in zip(EDGES[:-1], EDGES[1:])])
n_j = rng.poisson(THETA_TRUE * E_j)                    # 逐段觀測到的餘震數
alpha_t = ALPHA0 + np.concatenate([[0], np.cumsum(n_j)])
beta_t = BETA0 + np.concatenate([[0], np.cumsum(E_j)])

theta = np.linspace(0.02, 3.0, 500)
snap = [(0, "t = 0（通用先驗）"), (1, "0.25 天後"), (3, "1 天後"),
        (5, "7 天後"), (6, "30 天後")]
fig = go.Figure()
widths = []
for (k, lab), color in zip(snap, PALETTE):
    dist = gamma_dist(alpha_t[k], scale=1.0 / beta_t[k])
    widths.append(dist.ppf(0.95) - dist.ppf(0.05))
    fig.add_trace(go.Scatter(x=theta, y=dist.pdf(theta), mode="lines",
                             name=lab, line=dict(color=color, width=2.2)))
fig.add_vline(x=THETA_TRUE, line_dash="dash", line_color=QUAKE_COLOR,
              annotation_text=f"本序列真值 {THETA_TRUE:.2f}")
apply_layout(fig,
             title=f"貝氏後驗逐日變窄：90% 區間寬度從 {widths[0]:.2f} "
                   f"收斂到 {widths[-1]:.2f}（合成序列）",
             xaxis_title="生產力倍數 θ（1 = 區域通用值）",
             yaxis_title="後驗密度", hovermode="x", height=430)
fig
````

讀法有三層。**第一層**：藍色的先驗又寬又平，它老實承認「我對這
個序列一無所知，只知道歷史上的序列大致落在這個範圍」。**第二
層**：第一天結束時後驗已經收窄一大截，因為 Omori 律讓絕大多數
餘震擠在最初幾天——早期資料的資訊密度最高。**第三層**：後驗中心
從 1.0 一路移向 0.6，也就是系統逐漸發現「這個序列比通用值安靜」。

如果在第一天就照通用先驗發布，機率會系統性偏高。這不是理論上的
擔憂——下一節就是它真實發生的樣子。

## 22.4 上線後怎麼壞

2016 年紐西蘭 Kaikōura $M_w$ 7.8 是作業化系統最好的實戰教案，
因為它壞得非常典型。

序列初期，完整規模的估計出了問題。大震之後的頭幾小時到幾天，
小地震的波形被大地震與彼此的尾波淹沒，目錄的 $M_c$ 會急遽上升
再緩慢回落（{doc}`第 11 章 <11_catalog_completeness_b>`）。
Kaikōura 的自動估計沒能追上這個變化，$M_c$ 被固定在 3.95。門檻
一旦定死在高處，可用來擬合的餘震數就湊不到 100 個，第 22.2 節那
個切換機制便**卡在第一層**——系統只能一直用區域通用參數。

結果是：**通用預報高估了實際觀測到的餘震數**。

把責任歸位很重要。錯誤的源頭**不在模型公式**——R–J 與 STEP 的
數學一行都沒錯；也不在參數估計法——沒有資料，估計法再好也無米
可炊。源頭在**資料的即時品質**。第 14 章的 14.6 節已經證明過這
條因果鏈：不完整性會直接污染參數估計，$M_c$ 設錯會讓 $p$、$K$
系統性偏移。作業化系統只是把那個實驗室結論搬到了真實世界，並且
加上一個新的維度——**時間壓力**。研究時你可以等三個月，用重定位
過的目錄慢慢重估 $M_c$；作業系統必須在事件發生後幾分鐘內，用
一份還在滾動、還在補資料、還在改規模的即時目錄做決定。

這個落差本身就該寫進系統的風險清單。即時目錄與研究目錄的差別
不只是「精度差一點」：定位可能偏移數公里、規模可能差 0.2–0.3、
小事件會在數小時後被補進來（於是昨天的率被今天改寫）、偶爾還
會有非地震訊號混入。第 23 章會看到台灣的作業紀錄裡有同樣的
誠實記載——某段「平靜」無法僅憑即時目錄下定論。

**作業化系統的風險清單上，資料永遠排在模型前面**。

## 22.5 L'Aquila 與 ICEF

2009 年 4 月 6 日，義大利 L'Aquila 發生 $M_w$ 6.3 地震，約三百人
罹難。震前該區有持續數月的群震，官方留給公眾的訊息卻被廣泛理解
為「不會有大地震」。

後續的審判讓全世界的地震學家記住了一課。但這一課經常被記錯——
記成「說了會被告，所以最好不要說」。真相是雙面的：

```{admonition} L'Aquila 的兩面教訓
:class: warning
**一面**是說錯話的法律風險：科學家與官員因發布的訊息被追究責任，
這讓整個社群此後對「要不要公開講機率」變得緊張。
**另一面**卻方向相反：他們被批評的內容，恰恰是**沒有把「機率
確實升高了」講得夠清楚、夠強**。沉默的代價與說錯話的代價同樣
真實。只學到前一面的人，會得出「閉嘴最安全」這個完全錯誤的
結論。
```

這場災難直接催生了 **ICEF**（International Commission for
Earthquake Forecasting）與「作業化地震預報」這個領域
（Jordan et al. 2011）。ICEF 的建議可以濃縮成一句：**與其在
「能不能預測」上空轉、臨場即興發言，不如建立常態化、權威、
透明的機率發布制度**。

第 9 章給過 OEF 的定義，這裡補上它的來歷與代價。兩個關鍵詞值得
再看一次。**作業化（operational）**借自氣象學，指模型有自動、
及時的應用，因而資訊的「可行動性」提高。**權威（authoritative）**
是更硬的限定：資訊必須來自官方、可信、且**依法負有提供該資訊
之責**的來源。

這裡有一句話值得撐起一整節課：**權威性不能自我宣稱**。它必須由
更高層級（通常是政府）授予。一個研究團隊可以做出全世界最好的
模型、跑在最穩的機器上、通過所有 CSEP 檢驗，然後發現自己仍然
不能發布——因為第 5 步不在科學的管轄範圍內。這正是第 22.1 節
那道斷層最赤裸的樣子。

## 22.6 機率溝通是有實驗的科學

理工背景的讀者有一個很頑固的預設：**算出機率是科學，怎麼講是
公關**。這一節的目的就是拆掉它。機率溝通有假說、有對照實驗、
有可測量的結果變數——它是科學，而且是本章裡實驗證據最紮實的
一部分。

### 框架效應：一個介系詞就能改變理解

說「在未來一週**之內**（within the next week）」與說「在下一週
（in the next week）」，人們的推論不一樣：後者會讓相當比例的
受訪者以為事件將發生在**時窗的末端**（Doyle et al. 2020）。這是
可以設計實驗測出來的效應，不是文案品味。

更有效的做法是**換一個參照系**。Dryhurst et al.（2022）以跨國
實驗測試了多種表述，找出一個明顯改善理解的頻率框架：

> 想像有 100,000 個地震機率與此地完全相同的地方。在這一週，
> 我們預期其中 y 個地方會發生……、z 個地方不會發生……

把抽象的百分比翻譯成可數的地方數量，人的直覺就接得上了。用
Kaikōura 的真實預報數字把這個框架畫出來——每個點代表 1,000 個
「機率與此地相同」的地方：

````
# Kaikōura M7.8 後兩週（2016-11-28 起算）GeoNet 的未來 7 天預報
KAIKOURA = [("M5.0–5.9", 98, 5.6), ("M6.0–6.9", 41, 0.53),
            ("M≥7.0", 5, 0.05)]
N_DOTS = 100                             # 點陣的點數
PER_DOT = 100_000 // N_DOTS              # 每個點代表幾個「地方」
idx = np.arange(N_DOTS)
gx, gy = idx % 10, -(idx // 10)

fig = make_subplots(rows=1, cols=3, horizontal_spacing=0.06,
                    subplot_titles=[f"{lab}：{p}%（平均 {mu} 個）"
                                    for lab, p, mu in KAIKOURA])
for col, (lab, p, mu) in enumerate(KAIKOURA, start=1):
    hit = idx < p * N_DOTS // 100
    for mask, color, name in [(~hit, "#d5dde6", "不會發生"),
                              (hit, ACCENT, "會發生")]:
        fig.add_trace(go.Scatter(x=gx[mask], y=gy[mask], mode="markers",
                                 name=name, showlegend=(col == 1),
                                 marker=dict(size=11, color=color)),
                      row=1, col=col)
fig.update_xaxes(visible=False)
fig.update_yaxes(visible=False, scaleanchor="x")
apply_layout(fig,
             title=f"頻率框架：{N_DOTS * PER_DOT:,} 個「機率與此地相同」的"
                   f"地方裡，一週內會發生的有幾個"
                   f"（每點 = {PER_DOT:,} 個地方）",
             hovermode="closest", height=380)
fig
````

三張點陣圖來自**同一份預報**。左邊幾乎全滿，右邊只有五個點——
一份預報內部就跨了兩個數量級。這是「機率量級感」最有效的一張
圖：不需要任何統計訓練，也能一眼看出 98% 和 5% 是完全不同性質
的訊息。

順帶一個值得盯著看的算術。這三個規模帶同時公布了平均個數
5.6、0.53、0.05。若假設 Poisson，$P=1-e^{-\Lambda}$ 給出 41.1%
與 4.9%——與公布的 41% 與 5% **完全吻合**；但 5.6 個算出來是
99.6%，公布值卻是 98%。這不是筆誤，而是
{doc}`第 17 章 <17_testing_consistency>`負二項那一節的徵狀：
當率夠高時，序列之間的變異（過度離散）會讓「至少一次」的機率
**低於**同均值的 Poisson 值；率很低時這個效應是二階小量，所以
另外兩帶看不出來。**同一張表裡，兩種分布假設的差異剛好在最上面
那一格露出來**。

### 數量範圍比機率範圍抗扭曲

給不確定性範圍時有個兩難：有人只想要一個數字前面加「大約」，
有人想知道均值周圍的散布。但實務上有一個**可實證比較**的性質：
媒體轉述**機率區間**時幾乎只報最悲觀的上限；而轉述「預期地震
**數量**的範圍」（例如 0–2 個）卻相對忠實——2018 年阿拉斯加
M7.1 之後的經驗即是如此。

這條啟示很便宜也很有用：**同樣的不確定性，換一個載體，抗扭曲
能力就不一樣**。GeoNet 的預報表因此每一格同時給三個東西——平均
預期個數、可能範圍、至少發生一次的機率——再配上文字判讀語彙，
例如「29% 的機率（**unlikely**）」、「小於 1%（**extremely
unlikely**）」。數字給專業者，文字給所有人，範圍給媒體。

### 另外四條可操作的建議

這一節的每一條都來自實務評估或專家共識，而且都可以被抄。

- **把「已過期間的預報數」與「實際觀測數」並列**。看得到自己
  上一次說了什麼、後來真的發生了什麼，收訊者對預報準確性會
  安心得多（Becker et al. 2019）。這其實是把第 17 章的 N 檢驗
  從論文搬到公開網頁上——只是讀者是民眾。
- **分層提供資訊**。讓使用者自取所需的深度：最外層是一句
  非技術的訊息（例如「為更多地震做準備」），願意讀下去的人
  才進入表格、參數與方法說明。美國的事件頁就是這個結構——
  摘要、評註、預報表、模型參數四個分頁。
- **同理心訊息搭配整備建議**。純資訊的發布會讓收訊者停在焦慮
  上；接上「所以現在可以做什麼」才有出口。
- **常態化發布，而不是只在壞消息時出現**。只有壞消息才發言的
  機構，發言本身就變成了警報。

反過來，已經被辨識出來的溝通挑戰也有一張清單：預報會造成焦慮、
會影響觀光等產業；專業者與大眾都難以把機率轉成行動；大眾傾向
拿地震預報類比天氣預報（因而期待精確的時間地點）；大震沒有造成
明顯損害會產生**常態化偏誤**；以及最根本的一條——**人們想理解
不確定性，而我們還不知道怎麼有意義地傳達它**。

### 三情境：把機率翻譯成後果

Kaikōura 之後兩個多月，GeoNet 發布了一份涵蓋未來一年的**三情境
敘事**，被廣泛視為機率溝通的典範：

````
SCEN = [("情境一：餘震照預報衰減", 70, 1,
         "無新增重大災害；序列逐步平息"),
        ("情境二：M7.0–7.8", 25, 2,
         "區域性破壞；隱沒帶可能參與、局部海嘯"),
        ("情境三：大於主震", 5, 3,
         "廣域破壞；含 M8 以上板塊介面破裂的可能")]

fig = go.Figure()
for (lab, p, imp, desc), color in zip(SCEN, PALETTE):
    fig.add_trace(go.Scatter(x=[p], y=[imp], mode="markers+text",
                             name=f"{lab}（{p}%）", text=[f" {p}%"],
                             textposition="middle right",
                             marker=dict(size=18 + 34 * imp / 3, color=color,
                                         opacity=0.85),
                             hovertext=[desc], hoverinfo="text"))
fig.update_yaxes(tickvals=[1, 2, 3],
                 ticktext=["序列平息", "區域性破壞", "廣域破壞"],
                 title_text="潛在衝擊（序數，示意）", range=[0.4, 3.6])
apply_layout(fig,
             title=f"Kaikōura 三情境：機率合計 "
                   f"{sum(s[1] for s in SCEN)}%，且必有一個會發生",
             xaxis_title="未來一年的機率（%）", xaxis_range=[-4, 88],
             hovermode="closest", height=430, showlegend=True)
fig
````

這張圖的結構值得當成寫作模板。**橫軸是機率，縱軸是後果**——而
這正是「低機率、高衝擊」事件之所以難溝通的原因：兩個軸的方向
相反，人的注意力只會分配給其中一個。三情境的設計把兩者強制放在
同一頁上。

幾個關鍵設計要點：三個情境的機率**加起來是 100%**，並明講「其中
必有一個會發生」（避免讀者把三個當成三種可選的未來）；每個情境
同時給文字判讀、數字與潛在衝擊；並且明說「機率雖低，但比主震
之前高了許多」——把絕對機率與相對增益一起交代。使用者回饋還
追加了幾條：標示震度等級與可能觸發的次生災害（海嘯、液化）、
附上整備建議、解釋每個情境背後的**斷層行為**。這個結構與 COVID-19
疫情期間被廣泛使用的三情境傳播高度一致，看來不是巧合。

### 序列可以持續六年

溝通的另一個難處是**時間尺度**。學生（和多數民眾）對「餘震」的
直覺是幾天到幾週，但紐西蘭 Canterbury 序列給了一個非常不同的
圖像：2010 年 9 月 M7.1 Darfield 開場，2011 年 2 月 M6.2
Christchurch 造成 185 人罹難，而到了**第六年**的 2016 年 2 月，
一個 M5.7 仍然造成歷史建築立面損壞、停電、崖崩、落石與液化。

這次事件把「未來一年至少一個 M5.0–5.9」的機率從 **50%** 推到
**63%**。用合成序列把這條曲線畫出來——兩個目標值由建構決定，
其餘由 Omori 衰減自然給出：

````
T_DARFIELD, T_CHCH, T_M57 = 0.0, 0.47, 5.45      # 事件時間（年）
T_BEFORE = 5.075                                  # 2015-10-01 起算的一年
MU_BG, C_YR, P_YR = 0.05, 0.02, 1.10              # 背景率與 Omori（年）
P_BEFORE, P_AFTER = 0.50, 0.63                    # 公布的兩個機率


def _omori_int(t1, t2, t0):
    """單一母事件對 [t1, t2] 期望數的形狀因子。"""
    lo = max(t1, t0)
    return 0.0 if t2 <= lo else quad(
        lambda s: (s - t0 + C_YR) ** -P_YR, lo, t2)[0]


# 由 P(一年內至少一次) 反解兩個生產力：先解主序列，再解 M5.7
shape_b = (_omori_int(T_BEFORE, T_BEFORE + 1, T_DARFIELD)
           + 0.6 * _omori_int(T_BEFORE, T_BEFORE + 1, T_CHCH))
K_B = (-np.log(1 - P_BEFORE) - MU_BG) / shape_b
lam_base2 = (MU_BG + K_B * _omori_int(T_M57, T_M57 + 1, T_DARFIELD)
             + 0.6 * K_B * _omori_int(T_M57, T_M57 + 1, T_CHCH))
K_7 = (-np.log(1 - P_AFTER) - lam_base2) / _omori_int(T_M57, T_M57 + 1, T_M57)


def p_year(t):
    """在時刻 t 發布的「未來一年至少一次 M5.0–5.9」機率。"""
    lam = (MU_BG + K_B * _omori_int(t, t + 1, T_DARFIELD)
           + 0.6 * K_B * _omori_int(t, t + 1, T_CHCH))
    if t >= T_M57:
        lam += K_7 * _omori_int(t, t + 1, T_M57)
    return 1 - np.exp(-lam)


t_a = np.linspace(0.02, T_M57 - 1e-6, 160)
t_b = np.linspace(T_M57, 6.4, 100)
fig = go.Figure()
for tt, name, color in [(t_a, "M5.7 之前", ACCENT),
                        (t_b, "M5.7 之後", QUAKE_COLOR)]:
    fig.add_trace(go.Scatter(x=tt, y=[100 * p_year(s) for s in tt],
                             mode="lines", name=name,
                             line=dict(color=color, width=2.5)))
for t_ev, lab in [(T_DARFIELD, "M7.1"), (T_CHCH, "M6.2"), (T_M57, "M5.7")]:
    fig.add_vline(x=t_ev, line_dash="dot", line_color="#888888",
                  annotation_text=lab)
fig.add_trace(go.Scatter(x=[T_BEFORE, T_M57],
                         y=[100 * P_BEFORE, 100 * P_AFTER],
                         mode="markers", name="公布值（校準點）",
                         marker=dict(size=11, symbol="diamond",
                                     color="#4a3aa7")))
apply_layout(fig,
             title=f"序列第六年：一個 M5.7 把未來一年機率從 "
                   f"{100 * p_year(T_BEFORE):.0f}% 推到 "
                   f"{100 * p_year(T_M57):.0f}%（合成，校準至公布值）",
             xaxis_title="主震後時間（年）",
             yaxis_title="未來一年至少一次 M5.0–5.9 的機率（%）",
             hovermode="x", height=430)
fig
````

曲線的形狀本身就是訊息。**衰減很慢**——六年之後機率仍近五成，
因為 Omori 律是冪次而不是指數。**而且會被重新點燃**：一個比主震
小兩個規模單位的事件，就足以讓年機率跳升十幾個百分點。這兩件事
合起來，構成了「長期溝通」這個獨立的難題：**什麼時候可以停止
發布**？如何在六年裡維持適當程度的關注而不造成疲乏？新遷入的
居民怎麼辦？

### 最誠實的難題：降低驚訝，還是促成行動

Canterbury 序列第六年的工作坊研究，得到一個令人不安的發現：
**理解餘震序列行為的居民，在 M5.7 來襲時比較不驚訝——但也因此
沒有採取行動**。

這對整個溝通事業是根本性的挑戰。我們一直預設「讓人理解」就是
目的，但「不驚訝」與「有準備」是兩件事，甚至可能互斥：知道序列
會有活動尖峰的人，遇到尖峰時的反應是「果然來了」而不是「該做點
什麼」。

而那句最該記住的話來自另一群研究者：**人們要的是可行動的資訊，
不是機率**。這句話值得整個第二部反覆回味。我們花了十幾章學會
算機率——而機率只是中間產物。從機率走到可行動的資訊，中間還
需要衝擊模型、易損性模型、決策門檻（第 21 章的 OELF 就是這條
路的開頭）。**所以溝通不是把同樣的內容換個包裝，而是內容本身
必須改變**——這正是它是科學問題而非公關問題的證據。

## 22.7 專家共識與歧異（Delphi）

上面這些「應該怎麼做」的判斷，並不是某個人的意見。Mizrahi et
al.（2024）的回顧用 **Delphi 法**對 20 位國際專家做了正式的意見
徵集——匿名、多輪、每輪回饋群體結果讓專家有機會修正——目的是
分辨哪些是社群共識、哪些其實還在吵。

**有共識的部分**：

- **模型要納入高階餘震**（餘震的餘震）——這是唯一不分使用者
  群體都被認為重要的模型成分，也是 STEP 相對 ETAS 的已知弱點。
- **ETAS 應作為預設／基準模型**，並鼓勵使用集成。
- 檢驗必須**與基準模型比較**、必須是**前瞻或擬前瞻**、且應有
  **第三方參與**（例如 CSEP 實驗）。
- 「**開發者自己相信模型**」被明確判定為**不足夠**；模型可以
  更新參數，但**換一個區域必須重新檢驗**。
- 溝通產品**必須與使用者群體協同開發**——這是全部題目中共識
  最強的一條。

有一條共識則值得加註警語：專家目前**不認為** $b$ 值的時空變化
或大地測量資訊是重要的模型成分。回顧作者自己提醒，這只反映
**當下的研究狀態**——22.11 節要談的 FTLS 正是拿 $b$ 值時變做
判別，而近年也有用大地測量訊號做前兆的嘗試。專家共識是一張
快照，不是一條定律；把它當成研究方向的禁令，就誤讀了 Delphi
這個方法。

**沒有共識的部分**只有一條，但它是最關鍵的一條：**訂不出一組
「模型上線前必須通過」的最低檢驗集**。

為什麼訂不出來？回顧的分析很精采，因為分歧不在技術而在目的：

- **科學導向**的視角認為，應該有辦法確定性地淘汰太差的模型；
  檢驗的功能是**把關**。
- **使用者導向**的視角則認為，該依用途決定檢驗方式，甚至願意
  使用未經完整檢驗的模型——理由是：**如果替代方案是完全不給
  資訊，那麼即使簡單如 R–J 也能改善決策**。

兩種立場都不荒謬，而且無法用更多資料調解，因為它們對「檢驗是
為了什麼」有不同的答案。這是本章第二個「瓶頸不在統計」的例證。

還有一個很有意思的後設觀察：**專家在「溝通」議題上的共識，反而
比在「模型開發與檢驗」上高**。回顧作者的推測相當坦率——正因為
專家在溝通領域的經驗較少；在自己最專精的領域，每個人都堅持
自己的觀點。

## 22.8 沒有模型時怎麼辦：慢滑移與專家徵詢

Kaikōura 主震後第 10 天，GPS 資料顯示 Hikurangi 隱沒帶邊緣出現
**廣泛的慢滑移事件（SSE）**。這件事立刻帶來一個作業上的問題：
慢滑移會不會提高後續大地震的機率？提高多少？

問題在於，當時幾乎**沒有現成的研究可以依循**，更沒有任何一個
可以插上電就跑的模型。這正是作業化系統在真實世界裡最常遇到、
教科書卻最少討論的處境：**你需要一個數字，但沒有模型**。

GNS 的作法是**非結構化的專家徵詢**：請每一位專家獨立給出「未來
一年內中紐西蘭 $M\ge7.8$ 的機率」的最佳估計與 90% 信賴界限，再
把結果平均。得到的答案是 **5%（2%–8%）**。

這是全世界第一次為慢滑移事件做這種建模演練，值得記住的不是那個
5%，而是方法論上的一句話：**當模型不存在時，結構化（或至少有
紀律的）專家判斷本身就是方法**，不是打混。它可以被記錄、被
覆核、被事後檢驗，也可以誠實地把不確定性寫成一個區間。同樣的
邏輯在第 20 章出現過——複發模型的 logic tree 權重也來自專家
判斷；紐西蘭 Canterbury 的時變危害模型更進一步，用<strong>結構化
專家判斷（Cooke 古典模型）</strong>決定各模型的權重。差別只在於，那些
場合是為了選權重，這裡是為了憑空生出一個機率。

反過來說，它的極限也很清楚：專家徵詢**無法被前瞻檢驗**（一次性
事件，沒有樣本），也無法保證下一次徵詢會得到相同答案。它是
沒有模型時的臨時橋樑，不是模型的替代品。

## 22.9 倫理與開放科學

走到這裡，「這個模型該不該公開」已經不再是技術問題。地震學社群
近年把它明確地當成**倫理**問題來討論（Dallo et al. 2023），並且
分成兩層：

- **內部研究倫理**：資料保護、FAIR 原則、利益衝突、以及「產出
  開放科學」本身的義務。
- **外部研究倫理**：資訊誤用、對社會的責任、法律後果（L'Aquila
  訴訟被直接點名）、包容的研究文化。

外部倫理有兩個具體的風險特別值得放在心上。其一，**開放的地震
預報模型可能被誤解或誤用**，進而降低社會對模型的信任；其二，
更糟的是它可能被**刻意操弄以產生誇大的預報**，造成恐慌。開放
從來不是零成本的。

而開放的理由同樣硬。可重現性在這個領域不只是技術規範，而是
**倫理義務**——因為你的模型會影響別人的生命決策，所以你有義務
公開它，也有義務思考它被誤用的後果。實踐上的三件工具：
**可重現性套件**（程式碼＋資料＋執行環境，讓人不必額外努力就能
重現結果）、**FAIR 原則**（可尋、可取、可互通、可再用），以及
**pyCSEP**（開源的預報檢驗工具箱，讓人不必自己重造第 17、18 章
那整套檢驗架構）。

這裡與第 14 章接得非常緊。14.11 節指出，社群**缺一個真正標準化
的 ETAS 基準版本**——沒有兩個 ETAS 實作是相同的，於是「贏過
ETAS」這句話沒有明確意義。這正是「開放科學不只是道德訴求，而是
**科學進展的必要條件**」的最強論證：不能重現，就不能比較；不能
比較，這個領域就沒有淘汰機制。

最後是一個很少被寫進教科書的場景。那不勒斯的工作坊實際辯論過
「能不能就這樣把短期機率預報公開」，並且**投票**——多數支持
無條件公開，理由是「有些資訊對個人決策而言比沒有資訊更有用」
（與 22.7 節使用者導向的立場如出一轍）。但作者立刻補上兩句
限制。第一句我們已經見過：人們要的是可行動的資訊，不是機率。
第二句更尖銳：**用民主方式建立共識，這件事本身可能是不倫理
的**，因為在投票的房間裡，少數群體往往未被充分代表。

一個判準若連自己的正當性都要接受檢查，那大概是這個領域成熟的
徵兆。

## 22.10 常見誤解與陷阱

**誤解 1：「STEP 就是 USGS 現在在跑的系統。」** 不是。2005 年
上線的 STEP 即時系統早已走入歷史，美國現行的是**作業化餘震預報
（OAF）**，引擎是 R–J 或 ETAS，參數有通用、序列特定、貝氏結合
三種來源。這是很常見的世代誤植。附帶一提，紐西蘭的混成模型倒是
真的還在用 STEP 當短期元件之一——同一個名字在不同國家的現況
不同。

**誤解 2：「機率增益幾十倍，所以現在很危險。」** 增益大不等於
絕對機率高。第 14 章那條衰減曲線已經說得很清楚：主震剛過時
「一週內再來一個更大的」相對背景可以高上百倍，絕對值卻始終是
個位數到幾十個百分比。Kaikōura 的 $M\ge7$ 是 5%——比平時高了
幾個數量級，同時也是「二十次裡有十九次不會發生」。**兩句話都
要說，只說一句就是誤導**，而且是兩個方向的誤導。

**誤解 3：「取最大就是集成。」** 不是。Avmax 是**逐點取包絡**，
它不守恆——合成後的總期望事件數會**大於**任何單一模型
（附錄有兩行證明）。第 19 章的凸組合則守恆，而且只有加權平均
型的集成能藉由單一預報之間的離散度提供**認知不確定性**的資訊。
Avmax 是刻意保守的工程選擇，不是統計上的模型平均。

**誤解 4：「作業化系統的難點是模型精度。」** 本章從頭到尾都在
反駁這一條。Kaikōura 壞在 $M_c$ 估計（資料）；義大利不公開
（制度）；紐西蘭不預報 $M<5.0$（目錄歷史）；三情境為什麼要那樣
寫（社會科學）。模型公式在這些故事裡一行都沒錯。

**誤解 5：「餘震預報只在震後幾天有意義。」** Canterbury 序列
第六年的 M5.7 造成實質損害，並把年機率從 50% 推到 63%。Omori
律是冪次衰減——它的尾巴長到會出現在**行政與保險的時間尺度**上，
不只是新聞週期。

**誤解 6：「義大利有 OEF，所以義大利民眾看得到機率。」** 恰恰
相反。全世界唯一完整部署 OEF 的國家，是三國中**唯一不對大眾
公開**的。「有系統」與「有公開的資訊服務」是兩件事。

## 22.11 研究前沿與未解問題

**前震紅綠燈要上線還缺什麼。** FTLS（Foreshock Traffic Light
System, Gulia & Wiemer 2019）追蹤大震後 $b$ 值的時變，作為
「這是正常餘震序列，還是更大事件的前震序列」的一階判別：若
$b$ 值不升反降，燈號轉紅。原理有實驗室與野外支持，但要上線還
缺兩件事。第一，$b$ 值估計本身有多種偏差來源（$M_c$、規模
尺標、樣本數——第 11 章整章都在講這件事），而 FTLS 恰恰用在
$M_c$ 最不穩定的時段。第二，也是更根本的：**顏色警示不是機率
預報**，無法直接與 ETAS 或 STEP 做第 18 章那種資訊增益比較。
在轉換成機率之前，它不能進入集成，也不能被同一套標準檢驗。

**機器學習從卻步到重返。** 第 14 章的 14.11 節記錄了完整的故事
弧：DeVries et al.（2018）的深度學習餘震預測被 Mignan &
Broccardo（2019）證明與遠更簡單的模型資訊量相當，社群一度卻
步；近年 FERN 與神經點過程在資料充足時已能匹敵或略勝 ETAS，
而且**對目錄的非平穩性極為適應**——這正好對應本章 22.4 節的
痛點。真正的作業化考驗還在後面：它要能在半夜自動跑、在目錄
正在滾動時輸出、並且說得出自己為什麼給這個數字。

**衝擊型預報（impact-based forecasting）。** 從「地震發生機率」
走向「後果機率」——氣象服務的經驗顯示，提供衝擊資訊**能改善
風險感知，但不必然轉化為行動**。障礙是資料：缺乏衝擊資料、
需要理解受影響人群的易損性與應變能力，還要決定產品是為個人
還是社群尺度設計。

**OELF：把預報接到損失。** 這是第 21 章那條鏈的終點——義大利的
原型引擎每日接收全國地震率，配合建物清冊與易損性模型，換算成
預期的不可使用建物數、倒塌數與傷亡人數。紐西蘭的版本更貼近
工程決策：Canterbury 重建的時變危害模型，以及 Kaikōura 之後
支撐無筋磚造建築強制補強決策的機率增益計算。這條路上的每一步
都在增加需要的資料種類，也在增加被質疑的表面積。

**一個共同的天花板。** 所有現行作業化模型都是**叢集模型**，
因此都**無法以高機率預報未來的大地震**。它們的強項是餘震，不是
大地震。凡是宣稱能預報大地震的系統，先問它有沒有通過前瞻檢驗
與基準比較。

**台灣的制度選擇題。** 技術零件在台灣其實已經齊備（第 13、14
章的 ETAS 在地化與作業化快報）。剩下的是兩個非技術問題：機率
預報的**權威性由誰授予**？要走義大利路線（權威但不公開、先服務
民防）、紐西蘭路線（公開、與使用者共同設計）、還是美國路線
（只在大震後自動發布）？這題留給
{doc}`第 23 章 <23_taiwan_outlook>`正面回答。

## 22.12 附錄：Avmax 與相加的形式化比較

22.2 節說 STEP 在一個地方取最大、在另一個地方相加。把它寫成
兩行就不會再混淆。

**情況一：不同來源。** 設背景過程與觸發過程互不重疊地產生事件，
率分別為 $\lambda_{\rm bg}$ 與 $\lambda_{\rm cl}$。獨立 Poisson
過程的疊加仍是 Poisson，率相加：

$$\lambda(x) = \lambda_{\rm bg}(x) + \lambda_{\rm cl}(x),
  \qquad \Lambda = \Lambda_{\rm bg} + \Lambda_{\rm cl} .$$

這是正確的，因為兩者描述的是**互斥的事件族群**。

**情況二：同一族群的多個估計。** 設 $\lambda_1,\dots,\lambda_K$
都試圖描述**同一批**未來事件（例如多個母事件重疊的餘震區）。
若各自無偏，相加會得到
$\mathbb{E}[N] \approx K\,\mathbb{E}[N_{\rm true}]$——重複計數。
可行的合成有兩類——第 19 章的凸組合，以及 Avmax：

$$\begin{aligned}
\lambda_\pi(x) &= \sum_i \pi_i \lambda_i(x),
   \qquad \textstyle\sum_i \pi_i = 1 ,\\
\lambda_{\max}(x) &= \max_i \lambda_i(x) .
\end{aligned}$$

兩者的性質差很多。凸組合**守恆**：
$\Lambda_\pi = \sum_i \pi_i \Lambda_i$，總期望數落在各模型之間。
Avmax 則對每一點都滿足 $\lambda_{\max} \ge \lambda_i$，積分後

$$\Lambda_{\max} \;\ge\; \max_i \Lambda_i \;\ge\;
  \sum_i \pi_i \Lambda_i = \Lambda_\pi ,$$

第一個等號只在某一個 $\lambda_i$ **逐點支配**其他模型時成立。
換句話說，只要兩個模型的空間分布有交錯（實務上總是如此），
**Avmax 就會膨脹總期望事件數**。

這不是 bug，是設計。取最大等於在每一點採信最悲觀的那個模型，
是一個明確的保守選擇，適合危害度與工程決策的脈絡。但它的代價
要說清楚：$\Lambda_{\max}$ 會系統性地高於任何單一模型，因此
用第 17 章的 N 檢驗去驗它，**幾乎注定會被判定高估**；而且它
沒有權重，也就給不出第 19 章那個副產品——模型之間的離散度所
承載的認知不確定性。

一句話總結：**判準是「同一群還是不同群」，不是「保守還是不
保守」**；先分清族群，再決定要不要保守。
